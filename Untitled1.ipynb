```python
import pandas as pd

# 1. Load & Normalize Fields (CHANGE FILENAME BELOW IF NEEDED)
df = pd.read_csv("authentication-log-sample.csv")

# Standardize column names
df.columns = [c.strip().lower() for c in df.columns]
col_map = {}
for c in df.columns:
    if "time" in c or "date" in c:
        col_map[c] = "timestamp"
    elif "ip" in c or "source" in c:
        col_map[c] = "ip"
    elif "user" in c or "account" in c or "email" in c:
        col_map[c] = "account"
    elif "action" in c or "event" in c:
        col_map[c] = "action"
    elif "result" in c or "status" in c or "outcome" in c:
        col_map[c] = "result"
df = df.rename(columns=col_map)

# Normalize values
df["timestamp"] = pd.to_datetime(df["timestamp"], utc=True, errors="coerce")
df["ip"] = df["ip"].astype(str).str.strip()
df["account"] = df["account"].astype(str).str.strip().str.lower()
df["action"] = df["action"].astype(str).str.strip().str.lower()
df["result"] = df["result"].astype(str).str.strip().str.lower()
df = df.sort_values("timestamp").reset_index(drop=True)

print(f"Total Rows: {len(df)} | Start: {df['timestamp'].min()} | End: {df['timestamp'].max()}")

failures = df[df["result"].str.contains("fail|deny|invalid|error", na=False)]
successes = df[df["result"].str.contains("succ|allow|ok", na=False)]

# 2A. Password Spraying (1 IP hitting >= 4 accounts)
spray = (
    failures.groupby("ip")
    .agg(total_failures=("account", "count"), unique_accounts=("account", "nunique"),
         first_seen=("timestamp", "min"), last_seen=("timestamp", "max"))
    .query("unique_accounts >= 4")
)
print("\n=== 1. PASSWORD SPRAYING ===")
display(spray)

# 2B. Brute Force (1 Account with >= 5 failures)
brute = (
    failures.groupby("account")
    .agg(total_failures=("ip", "count"), unique_ips=("ip", "nunique"),
         first_seen=("timestamp", "min"), last_seen=("timestamp", "max"))
    .query("total_failures >= 5")
)
print("\n=== 2. BRUTE FORCE ===")
display(brute)

# 2C. Unusual Access
suspicious_ips = set(spray.index).union(set(failures["ip"].value_counts()[lambda x: x >= 3].index))
unusual_success = successes[
    (successes["ip"].isin(suspicious_ips)) | (successes["timestamp"].dt.hour.between(0, 5))
]
print("\n=== 3. UNUSUAL ACCESS ===")
display(unusual_success)

# 3. Full Timeline
flagged_accounts = set(brute.index).union(set(unusual_success["account"]))
timeline = df[(df["ip"].isin(suspicious_ips)) | (df["account"].isin(flagged_accounts))]
print("\n=== 4. INCIDENT TIMELINE ===")
display(timeline)
```
